```python
# Import pandas for data manipulation
import pandas as pd

# Import Path to handle file paths cleanly
from pathlib import Path
```

```python
# Path containing the original monthly BTS flight files
raw_path = Path("../data/raw/flights")

# Path where the combined dataset will be saved
processed_path = Path("../data/processed")

# Create processed folder if it does not already exist
processed_path.mkdir(parents=True, exist_ok=True)
```

```python
# Read January 2026 BTS flight data
jan = pd.read_csv(raw_path / "T_ONTIME_MARKETING jan.csv")

# Display the number of rows and columns
print("January:", jan.shape)
```

```python
# Read February 2026 BTS flight data
feb = pd.read_csv(raw_path / "T_ONTIME_MARKETING feb.csv")

# Display the number of rows and columns
print("February:", feb.shape)
```

```python
# Read March 2026 BTS flight data
march = pd.read_csv(raw_path / "T_ONTIME_MARKETING march.csv")

# Display the number of rows and columns
print("March:", march.shape)
```

```python
# Check whether all three monthly datasets have exactly the same columns
print("January == February:",
      list(jan.columns) == list(feb.columns))

print("February == March:",
      list(feb.columns) == list(march.columns))
```

```python
# Find columns that exist in March but not in February
march_extra = set(march.columns) - set(feb.columns)

# Find columns that exist in February but not in March
feb_extra = set(feb.columns) - set(march.columns)

print("Columns only in March:")
print(march_extra)

print("\nColumns only in February:")
print(feb_extra)
```

```python
# Compare the number of columns
print("February columns:", len(feb.columns))
print("March columns:", len(march.columns))

# Show columns side by side
for i, (feb_col, march_col) in enumerate(zip(feb.columns, march.columns)):
    if feb_col != march_col:
        print(
            f"Position {i}: "
            f"February = {feb_col} | "
            f"March = {march_col}"
        )
```

```python
# Combine the three monthly datasets vertically
# ignore_index=True creates a new continuous index
flights = pd.concat(
    [jan, feb, march],
    ignore_index=True
)

# Display the combined dataset size
print("Combined dataset shape:", flights.shape)
```

```python
# Count the number of records belonging to each month
print(
    flights["MONTH"]
    .value_counts()
    .sort_index()
)
```

```python
# Convert FL_DATE to a proper datetime column.
# format='mixed' allows Pandas to handle different string representations.
flights["FL_DATE"] = pd.to_datetime(
    flights["FL_DATE"],
    format="mixed",
    errors="coerce"
)
```

```python
sample=flights.head(100000)
```

```python
# Save the combined dataset as a new CSV.
# The original monthly BTS files remain untouched.
output_file = processed_path / "flights_2026_q1.csv"

sample.to_csv(output_file, index=False)

print(f"Combined dataset saved to: {output_file}")
```

```python
# Read the combined file back from disk
flights_check = pd.read_csv(output_file)

# Verify its shape
print("Saved dataset shape:", flights_check.shape)

# Verify its date range
print("Date range:",
      flights_check["FL_DATE"].min(),
      "to",
      flights_check["FL_DATE"].max())
```
