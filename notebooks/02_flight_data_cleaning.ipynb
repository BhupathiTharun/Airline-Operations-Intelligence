# PART 7 — VALIDATE HHMM TIME COLUMNS

`dep_time` and `arr_time` are stored as HHMM-style numbers.

Examples:

- `530` → 05:30
- `1345` → 13:45
- `2460` → invalid
- `1265` → invalid

We need to check the values before cleaning them.

```python
df.head()
```

```python
def valid_hhmm(value):
    if pd.isna(value):
        return True

    try:
        value = int(value)
    except:
        return False

    hour = value // 100
    minute = value % 100

    return (
        0 <= hour <= 23
        and 0 <= minute <= 59
    )
```

```python
invalid_dep_time = (
    df["dep_time"].notna()
    & ~df["dep_time"].apply(valid_hhmm)
)

print(
    "Invalid dep_time values:",
    invalid_dep_time.sum()
)
df[invalid_dep_time].head(20)
```

```python
invalid_arr_time = (
    df["arr_time"].notna()
    & ~df["arr_time"].apply(valid_hhmm)
)

print(
    "Invalid arr_time values:",
    invalid_arr_time.sum()
)
df[invalid_arr_time].head(20)
```

```python
bad_dep_records = df.loc[
    invalid_dep_time,
    [
        "fl_date",
        "mkt_carrier",
        "mkt_carrier_fl_num",
        "origin",
        "dest",
        "crs_dep_time",
        "dep_time",
        "dep_delay"
    ]
]

display(bad_dep_records.head(20))
```

# PART 12 — CLEANING DECISIONS

| Problem | Cleaning decision |
|---|---|
| Duplicate rows | Remove only if duplicates exist |
| Invalid HHMM | Replace invalid value with `NaN` |
| Invalid numeric measurement | Replace invalid value with `NaN` |
| Cancelled-flight structural missingness | Keep as `NaN` |
| Suspicious completed-flight measurement | Keep record + create quality flag |
| Extreme delays | Keep record + create outlier flag |

The important principle is:

**Do not delete an entire flight just because one measurement is wrong.**
