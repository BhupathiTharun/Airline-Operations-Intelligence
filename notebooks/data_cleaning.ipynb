```python
def hhmm_to_minutes(x):
    h = int(x) // 100
    m = int(x) % 100
    return h * 60 + m

test = suspicious[
    ['crs_dep_time', 'dep_time', 'dep_delay']
].copy()

test['crs_dep_min'] = test['crs_dep_time'].apply(hhmm_to_minutes)
test['dep_min'] = test['dep_time'].apply(hhmm_to_minutes)

test['calculated_dep_delay'] = (
    test['dep_min'] - test['crs_dep_min']
)

# Handle flights crossing midnight
test.loc[
    test['calculated_dep_delay'] < -720,
    'calculated_dep_delay'
] += 1440

test[['crs_dep_time', 'dep_time', 'dep_delay',
      'calculated_dep_delay']]
```

```python
check = suspicious[
    ['crs_arr_time', 'arr_time', 'arr_delay',
     'arr_delay_new', 'arr_del15']
].copy()

check
```

```python
def calculate_arr_delay(row):
    if pd.isna(row['arr_time']) or pd.isna(row['crs_arr_time']):
        return np.nan

    sched = int(row['crs_arr_time'])
    actual = int(row['arr_time'])

    sched_h, sched_m = divmod(sched, 100)
    actual_h, actual_m = divmod(actual, 100)

    sched_minutes = sched_h * 60 + sched_m
    actual_minutes = actual_h * 60 + actual_m

    delay = actual_minutes - sched_minutes

    # Handle midnight crossing
    if delay < -720:
        delay += 1440
    elif delay > 720:
        delay -= 1440

    return delay


check['calculated_arr_delay'] = check.apply(
    calculate_arr_delay,
    axis=1
)

check['difference'] = (
    check['arr_delay'] - check['calculated_arr_delay']
)

check
```

```python
checks = {
    'Cancelled flights with dep_time':
        ((df['cancelled'] == 1) & df['dep_time'].notna()).sum(),

    'Cancelled flights with arr_time':
        ((df['cancelled'] == 1) & df['arr_time'].notna()).sum(),

    'Diverted + Cancelled':
        ((df['diverted'] == 1) & (df['cancelled'] == 1)).sum(),

    'Invalid taxi_out':
        ((df['taxi_out'] < 0) | (df['taxi_out'] > 300)).sum(),

    'Invalid taxi_in':
        ((df['taxi_in'] < 0) | (df['taxi_in'] > 300)).sum(),

    'Invalid CRS elapsed':
        ((df['crs_elapsed_time'] <= 0) |
         (df['crs_elapsed_time'] > 1000)).sum()
}

pd.Series(checks)
```

```python
# ==============================
# BATCH NUMERIC VALIDATION
# ==============================

rules = {
    'taxi_out': (0, 300),
    'taxi_in': (0, 300),
    'crs_elapsed_time': (1, 1000),
    'actual_elapsed_time': (1, 1500),
    'air_time': (1, 1500),
    'distance': (1, 6000),
}

validation = {}

for col, (min_val, max_val) in rules.items():
    validation[col] = (
        ((df[col] < min_val) | (df[col] > max_val)).sum()
    )

pd.Series(validation, name='invalid_count')
```
