```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    r"D:\Data Analyst\EXCEL\api\processed\flights_2026_q1_cleaned.csv"
)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

# KPI Baseline

```python
df['fl_date']=pd.to_datetime(df['fl_date'])
```

### Total Flights

```python
total_flights = len(df)

print("Total Flights:", f"{total_flights:,}")
```

### Cancelled Flights

```python
cancelled_flights = (df['cancelled'] == 1).sum()
print("Cancelled Flights:", cancelled_flights)
```

### Cancellation Rate

```python
cancellation_rate = (
    cancelled_flights / total_flights
) * 100

print(
    "Cancellation Rate:",
    f"{cancellation_rate:.2f}%"
)
```

### Diverted Flights

```python
diverted_flights = df["diverted"].eq(1).sum()

print(
    "Diverted Flights:",
    f"{diverted_flights:,}"
)
```

```python
diversion_rate = (
    diverted_flights / total_flights
) * 100

print(
    "Diversion Rate:",
    f"{diversion_rate:.2f}%"
)
```

### Departure Delay

```python
dep_delay_available = df["dep_delay"].notna().sum()

print(
    "Flights with departure delay data:",
    f"{dep_delay_available:,}"
)
```

```python
median_dep_delay = df["dep_delay"].median()

print(
    "Median Departure Delay:",
    f"{median_dep_delay:.1f} minutes"
)
```

```python
avg_dep_delay = df["dep_delay"].mean()

print(
    "Average Departure Delay:",
    f"{avg_dep_delay:.1f} minutes"
)
```

### Arrival Delay

```python
arr_delay_available = df["arr_delay"].notna().sum()

print(
    "Flights with arrival delay data:",
    f"{arr_delay_available:,}"
)
```

```python
median_arr_delay = df["arr_delay"].median()

print(
    "Median Arrival Delay:",
    f"{median_arr_delay:.1f} minutes"
)
```

```python
avg_arr_delay = df["arr_delay"].mean()

print(
    "Average Arrival Delay:",
    f"{avg_arr_delay:.1f} minutes"
)
```

### On-Time Arrival Rate

```python
eligible_arrival = df[
    (df["cancelled"] == 0) &
    (df["diverted"] == 0) &
    (df["arr_delay"].notna())
].copy()
```

```python
print(
    "Eligible arrival flights:",
    f"{len(eligible_arrival):,}"
)
```

```python
on_time_arrival_rate = (
    eligible_arrival["arr_delay"].le(15).mean()
) * 100

print(
    "On-Time Arrival Rate:",
    f"{on_time_arrival_rate:.2f}%"
)
```

### Airline × Arrival Delay

```python
airline_delay = (
    df.groupby("mkt_unique_carrier")
      .agg(
          flights=("mkt_unique_carrier", "size"),
          median_arr_delay=("arr_delay", "median"),
          avg_arr_delay=("arr_delay", "mean")
      )
      .sort_values("median_arr_delay", ascending=False)
)

airline_delay
```

### Airline × Cancellation Rate

```python
airline_cancel = (
    df.groupby("mkt_unique_carrier")
      .agg(
          flights=("mkt_unique_carrier", "size"),
          cancelled=("cancelled", "sum")
      )
)

airline_cancel["cancellation_rate"] = (
    airline_cancel["cancelled"] /
    airline_cancel["flights"] * 100
)

airline_cancel.sort_values(
    "cancellation_rate",
    ascending=False
)
```

```python
plt.figure(figsize=(10, 6))

airline_cancel["cancellation_rate"].sort_values().plot(
    kind="barh"
)

plt.title("Cancellation Rate by Airline")
plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Airline")

plt.show()
```

### Volume vs Performance

```python
airline_performance = (
    df.groupby("mkt_unique_carrier")
      .agg(
          flights=("mkt_unique_carrier", "size"),
          median_arr_delay=("arr_delay", "median"),
          cancellation_rate=("cancelled", "mean")
      )
)

airline_performance["cancellation_rate"] *= 100

airline_performance
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    airline_performance["flights"],
    airline_performance["median_arr_delay"]
)

for airline, row in airline_performance.iterrows():
    plt.annotate(
        airline,
        (row["flights"], row["median_arr_delay"])
    )

plt.title("Airline Volume vs Median Arrival Delay")
plt.xlabel("Number of Flights")
plt.ylabel("Median Arrival Delay (minutes)")

plt.show()
```

# Multivariate

```python
delay_causes = [
    "carrier_delay",
    "weather_delay",
    "nas_delay",
    "security_delay",
    "late_aircraft_delay"
]
```

```python
df[delay_causes].describe().T
```

```python
cause_totals = (
    df[delay_causes]
    .sum()
    .sort_values(ascending=False)
)

cause_totals
```

```python
plt.figure(figsize=(10, 6))

cause_totals.sort_values().plot(
    kind="barh"
)

plt.title("Total Delay Minutes by Cause")
plt.xlabel("Total Delay Minutes")
plt.ylabel("Delay Cause")

plt.show()
```

```python
cause_contribution = (
    cause_totals /
    cause_totals.sum() * 100
)

cause_contribution
```

```python
plt.figure(figsize=(10, 6))

cause_contribution.sort_values().plot(
    kind="barh"
)

plt.title("Share of Total Delay by Cause")
plt.xlabel("Share of Delay (%)")
plt.ylabel("Delay Cause")

plt.show()
```

```python
airline_causes = (
    df.groupby("mkt_unique_carrier")[delay_causes]
      .sum()
)

airline_causes
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    airline_causes,
    annot=True,
    fmt=".0f"
)

plt.title("Total Delay Minutes by Airline and Cause")
plt.xlabel("Delay Cause")
plt.ylabel("Airline")

plt.show()
```

```python
final_kpi = pd.Series({
    "Total Flights": len(df),
    "Cancelled Flights": df["cancelled"].eq(1).sum(),
    "Cancellation Rate (%)": df["cancelled"].mean() * 100,
    "Diverted Flights": df["diverted"].eq(1).sum(),
    "Diversion Rate (%)": df["diverted"].mean() * 100,
    "Median Departure Delay (min)": df["dep_delay"].median(),
    "Median Arrival Delay (min)": df["arr_delay"].median(),
    "Average Arrival Delay (min)": df["arr_delay"].mean(),
    "On-Time Arrival Rate (%)": (
        eligible_arrival["arr_delay"].le(15).mean() * 100
    ),
    "Delay Minutes / Flight": (
        df["arr_delay"].sum() / df["arr_delay"].notna().sum()
    )
})

final_kpi
```

```python
kpi_definitions = {
    "Cancellation Rate":
        "Cancelled flights / total flights",

    "Diversion Rate":
        "Diverted flights / total flights",

    "Median Arrival Delay":
        "Median arrival delay among flights with valid arrival-delay data",

    "On-Time Arrival Rate":
        "Eligible completed flights arriving within 15 minutes of schedule",

    "Delay Minutes / Flight":
        "Total recorded arrival-delay minutes / flights with arrival-delay data"
}
```

```python
airline_final = (
    df.groupby("mkt_unique_carrier")
      .agg(
          flights=("mkt_unique_carrier", "size"),
          median_arr_delay=("arr_delay", "median"),
          cancellation_rate=("cancelled", "mean"),
          diversion_rate=("diverted", "mean")
      )
)

airline_final["cancellation_rate"] *= 100
airline_final["diversion_rate"] *= 100

airline_final["delay_minutes_per_flight"] = (
    df.groupby("mkt_unique_carrier")["arr_delay"].sum()
    /
    df.groupby("mkt_unique_carrier")["arr_delay"].count()
)

airline_final
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    airline_final["cancellation_rate"],
    airline_final["median_arr_delay"]
)

for airline, row in airline_final.iterrows():
    plt.annotate(
        airline,
        (
            row["cancellation_rate"],
            row["median_arr_delay"]
        )
    )

plt.title(
    "Airline Operational Risk: Cancellation vs Arrival Delay"
)

plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Median Arrival Delay (minutes)")

plt.show()
```
