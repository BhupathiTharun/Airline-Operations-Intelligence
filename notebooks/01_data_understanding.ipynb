## 1. Business Context

Flight delays, cancellations, and diversions can affect airline operations,
airport capacity, passenger experience, and overall network performance.

This project aims to understand the operational patterns behind flight
disruptions and identify the airlines, airports, routes, and time periods
where disruption is more common.

The analysis will later combine flight operations data with airport reference
and historical weather data.

## 2. Business Questions

The analysis will eventually answer questions such as:

1. What is the overall flight delay rate?
2. Which airlines have the highest delay rates?
3. Which airports experience the most delays?
4. Which routes have higher disruption rates?
5. Does the time of day affect delays?
6. Does airport congestion relate to delays?
7. Does weather relate to flight delays?
8. What are the major causes of delays?
9. Can high-risk flights be identified before departure?
