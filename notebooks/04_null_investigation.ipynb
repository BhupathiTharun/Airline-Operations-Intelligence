```python
cancelled=df[df['cancelled']==1][['dep_time','dep_delay', 'dep_delay_new', 'dep_del15', 
    'taxi_out', 'taxi_in', 'arr_time', 'arr_delay', 'arr_delay_new', 'arr_del15',
    'cancelled','actual_elapsed_time', 'air_time', 'carrier_delay',
       'weather_delay', 'nas_delay', 'security_delay', 'late_aircraft_delay']]
cancelled
```

```python
cols = [
    'dep_time', 'dep_delay', 'dep_delay_new', 'dep_del15',
    'taxi_out', 'taxi_in', 'arr_time', 'arr_delay',
    'arr_delay_new', 'arr_del15', 'actual_elapsed_time',
    'air_time', 'carrier_delay', 'weather_delay',
    'nas_delay', 'security_delay', 'late_aircraft_delay'
]

# Find cancelled flights where all these columns are NULL
mask = (
    (df['cancelled'] == 1) &
    (df[cols].isna().all(axis=1))
)

# Convert columns to object so they can contain text
df[cols] = df[cols].astype('object')

# Fill only those cancelled rows
df.loc[mask, cols] = 'flight cancelled'
```
