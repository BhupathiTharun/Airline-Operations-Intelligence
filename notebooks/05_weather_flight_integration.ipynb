```python
def get_weather_category(code):

    if code == 0:
        return "Clear"

    elif code in [1, 2, 3]:
        return "Cloudy"

    elif code in [51, 53, 55, 56, 57, 61, 63, 65, 66, 67, 80, 81, 82]:
        return "Rain"

    elif code in [71, 73, 75, 77, 85, 86]:
        return "Snow"

    elif code in [95, 96, 99]:
        return "Storm"

    else:
        return "Unknown"
```

```python
flights_enriched["weather_category"] = (
    flights_enriched["weather_code"].apply(get_weather_category)
)
```
