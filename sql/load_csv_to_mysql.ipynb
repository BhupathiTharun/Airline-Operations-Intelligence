```python
import csv
import mysql.connector

# 1. Connect to Database
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Tharun8292@",
    port=3307,
    database="airline_intelligence",
)
cur = db.cursor()

# 2. Open CSV file
path = r"D:\Data Analyst\EXCEL\airline-operations-intelligence\Airline-Operations-Intelligence\data\processed\flights_2026_q1_cleaned_sql.csv"
file = open(path, mode="r", encoding="utf-8")
reader = csv.reader(file)

# 3. Get headers and create table
headers = next(reader)
columns_sql = ", ".join([f"`{h}` TEXT" for h in headers])
cur.execute(f"CREATE TABLE IF NOT EXISTS flights_2026_q1 ({columns_sql})")

# 4. Insert data in chunks
placeholders = ", ".join(["%s"] * len(headers))
insert_sql = f"INSERT INTO flights_2026_q1 VALUES ({placeholders})"

chunk_size = 5000
batch = []

for row in reader:
    batch.append(row)

    # When batch reaches 5000 rows, insert and clear
    if len(batch) == chunk_size:
        cur.executemany(insert_sql, batch)
        db.commit()
        batch = []

# Insert any leftover rows
if batch:
    cur.executemany(insert_sql, batch)
    db.commit()

# 5. Close connections
file.close()
cur.close()
db.close()

print("All chunks uploaded successfully!")
```
